This notebook demonstrates how to:
- Chunk text for embedding
- Generate embeddings using a local HuggingFace model
- Save and load vector index locally

```python
from config import settings
```

```python
def chunk_text(text, chunk_size=500, overlap=50):
    chunks = []
    start = 0
    while start < len(text):
        end = start + chunk_size
        chunks.append(text[start:end])
        start = end - overlap
    return chunks
```

```python
text = "hhiutibut7utuytv yuvtuyrt uyr yu ryt ryrruyr" ## settings.PROCESSED_FILE

chunked_text = chunk_text(
    text,
    settings.SENTENCE_TRANSFORMER_CHUNK_SIZE,
    settings.SENTENCE_TRANSFORMER_OVERLAP_SIZE,
)

print(chunked_text)
```
